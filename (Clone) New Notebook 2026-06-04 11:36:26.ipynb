```python
# Read CSV file from Volume
df = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load("/Volumes/helathcare/bronze/patients/*.csv")
    
display(df.count())
```

```python
df.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable("helathcare.bronze.patent_details")
```

```python
spark.table("helathcare.bronze.patent_details").display()
```

```python
# OPTIMIZE Delta table
spark.sql("""
OPTIMIZE helathcare.bronze.patent_details
""")
display(spark.table("helathcare.bronze.patent_details"))

```

```python
# VACUUM Delta table
spark.sql("""
VACUUM helathcare.bronze.patent_details RETAIN 168 HOURS
""")
```

```python
%sql
DESCRIBE history helathcare.bronze.patent_details
```
